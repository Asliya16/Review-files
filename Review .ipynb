## Q2 

```python
daily_steps={
    "mon":8500,
    "tue":12000,
    "wed":6200,
    "thu":10500,
    "fri":9800,
    "sat":15000,
    "sun":4300}
```

```python
total =0
goal_days=0
```

```python
for day in daily_steps:
    total += daily_steps[day]
    if daily_steps[day] >= 10000:
        print (day, ":",daily_steps[day],"goal acheived")
        goal_days += 1
    else:
        print(day, ":",daily_steps[day],"below goal")
```

```python
print("goal achieved:",goal_days,"days")
average =total /7
print("weekly average:",round (average,2),"steps")
```
